## Q6. Median value of horsepower


```python
# Step 1: Median before filling
median_before = df['horsepower'].median()
print(f"Median horsepower (before fillna): {median_before}")
```

```python
# Step 2: Most frequent value (mode)
mode_hp = df['horsepower'].mode()[0]
print(f"Mode of horsepower: {mode_hp}")
```

```python
# Step 3: Fill missing values with mode
df_filled = df.copy()
df_filled['horsepower'] = df_filled['horsepower'].fillna(mode_hp)
```

```python
# Step 4: Median after filling
median_after = df_filled['horsepower'].median()
print(f"Median horsepower (after fillna): {median_after}")
print(f"\nBefore: {median_before}  →  After: {median_after}")
if median_after == median_before:
    print("Result: No change")
elif median_after > median_before:
    print("Result: Yes, it increased")
else:
    print("Result: Yes, it decreased")
```

**Answer :** Yes, it decreased (254.0 → 252.0)

## Q7. Sum of weights



```python
# Step 1: Select cars from Asia
asia_cars = df[df['origin'] == 'Asia']

# Step 2: Select columns
subset = asia_cars[['vehicle_weight', 'model_year']]

# Step 3: First 7 values
subset = subset.head(7)
print("First 7 Asian cars (vehicle_weight, model_year):")
subset
```

```python
# Step 4: Get NumPy array
X = subset.values
print("X =")
print(X)
```

```python
# Step 5: Compute XTX (matrix-matrix multiplication)
XTX = X.T.dot(X)
print("XTX =")
print(XTX)
```

```python
# Step 6: Invert XTX
XTX_inv = np.linalg.inv(XTX)
print("XTX inverse =")
print(XTX_inv)
```

```python
# Step 7: Create y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
print("y =", y)
```

```python
# Step 8: Compute w = (XTX)^-1 @ X^T @ y
w = XTX_inv.dot(X.T).dot(y)
print("w =", w)
```

```python
# Step 9: Sum of all elements of w
print(f"Sum of w: {w.sum():.4f}")
```

**Answer Q7:** `0.369` (sum ≈ 0.3692)
